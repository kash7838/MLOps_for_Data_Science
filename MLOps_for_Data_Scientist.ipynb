# Machine Learning Operations - MLOps

### Problem Statement
- Building Fraud Detecting System for a banking application.
- It needs to flag suspicious transaction to protect customers from fraud.
- Reviewing a trend of transaction that don't match your normal transaction.
- Goal: 95% of times keeping fraudulent transactions low.

### Scenario
- After completing the project -> Data Scientists deployed the model it on production server.
- But the model started to break down. The Data Scientists built their model with Python but the bank uses Java.
- Java provides stability, security, performance as well as enterprise support for transaction heavy workloads.
- The use of python can't be implement because of compliance, security policy, lack of python support or resource constraints.
- Hence they have to re-write the logic in Java.
- Now, suppose if every transaction is taking 3 seconds to review and analyze then it won't work for bank because every minute bank might have thousands, hundred of thousands, or millions of transactions which needed to get review. This solution will be declared a failure.
- Performance mismatch is a common pain point which we might see. Prototypes working locally well but not on production level. This can happen because of not optimising code or environment mismatched. 
- If the model is built and tested on laptop but deployed on cloud such as AWS or Kuberenetes clusters, this may lead to mismatch of system libs, dependency versions or Available memory or CPUs/GPUS, which will lead to errors.
- After a month or weeks, the model starts to missing many fraud cases because of the arrival of new patterns for which the ML model is not prepared -> Data Drift Problem. 
- Because of lack of documentation, even after providing new updated example or logic, the performance is not upto date because no one knows how the model was built originally which leads to reproducibility.
- And the error also was only highlighted when the customer flagged or complained about this issue. There was no system monitoring the model's actively, whether it was working correctly or not.  

### Solution we need
- Need of consistent working environment:  
    - It should perform consistently same on development server, staging server or prodcution server the way it does on Data Scientist laptop.
    - Data -> Steps -> Results should be reliable and predictable wherever the models runs, whether it runs on laptop or banks live system. Different results, Unexpected errors, failures cause by environment changes can be avoided and the team trust the model at every stage.

- Performance Testing:
    - We need to check the performance before putting it front of real customers. 
    - Example: Model is used to evaluate how risky the transaction is, team should measure how quickly it process a transaction, whether it takes a second or milliseconds, this way team will know ahead of time if the model's speed be able to keep up in the real world demands.

- Data Monitoring:
    - It is important to make sure that the data coming in looks like the data we used for training.
    - If the customer suddenly making transaction huge amounts at night time which used to be rare. The team should know this quickly, it will help the team to get update on new patterns on customer behavior or new fraud patterns right away, so, the model can be updated before it's results become unreliable.

- Documentation:
    - We need to document every detail of how we built our model.
    - What data we used, what steps we took to clean it, what settings we choose and so on.
    - Example: If a team is training a fraud detection model using a customer transaction data from 2024: Q1 & Q2 -> applied filter such as remove duplicates entries and added a condition that flags transfer over 8000$ as *high risk*. All of these things should be recorded, this way anyone or team always re-build or review the model exactly the same in future. Making sure that reuslts are always reproducible and easy to audit.

- Real Time Monitoring Dashboard:
    - We need a dashboard that shows how well our model is performing at all times. 
    - If the model suddenly starts making mistakes such as missing actual fraud or incorrectly flagging safe transactions.
    - Dashboard should highlight the increase in errors and send an alert to the team right away. Then the team can quickly identify and fix the problem before they impact customer or business operations.

- Model Updates:
    - Easy way to update our model when we needed without disrupting the bank's service or suspending for a defined period of time.
    - If new patterns of fraud appears or customer behavior changes, the team should be able to rollout a better model quickly, so, the change happens smoothly with no downtime of the service and customers keep using the bank's services as normal.

### MLOps 
- It helps us with all of this situations mentioned above. MLOps stands for Machine Learning Operations.
    - It gets models from a Data Scientist model local device(laptop or PC) into real production environment.
    - Packaging them properly setting them up to run reliably where your customer's actually use them.
    - It also helps us keep those models running correctly -> ML models need someone watching over them -> MLOps gives you ways to check if your model is doing good job, fix problem when it occurs, keep everything run smoothly.
    - It helps us update models easily when needed. When you have new data or things changed, MLOps lets you swap in new model without breaking everything or having downtimes.
    - It's almost looking fueling an aeroplane when it is flying.

- Example of Fraud Detection System:
    - With good MLOps, the system automatically watches if the fraud catching model is working properly.
    - If the model start mixing fraud cases -> it will quickly alerts the team.
    - Team will quickly update the model with examples of the new fraud or switch back to the old model that was working better.
    - Without causing frication to the bank or the customer using that service.

- Another way you can look at MLOps will be what MLOps is for machine learning that's how a kitchen is to cooking. Even if we have the best data scientist or best chef, they won't be able to seve unless they a proper set of tools along with ingredients

### Need of MLOps 
- Tradition software is like building a house with a detailed blueprint. Once built it works as designed unless you change the code. But ML systems are more like building a house that continously change itself based on weather, occupants, preferences,neighborhood changes, etc. 
- ML Models have unique challenges as compared to software applications. They heavily depend on data which constantly changes in the real world. 
- Their performace can decrease over time called Model Drift.
- They need continous retraining with new data. The process may involved specialist like Data Scientist, Data Enigneer, AI engineer, ML engineer or SMEs.
- Netflix uses ML models to recommend content to its millions of users. If they built a model once and deploy it wihtout MLOps practices. The content will become less relevant over time as new content is added and user prefernces changes.
- Netlfix needs systems to continously collect the viewing data, retrain their models and deploy the updated versions.              

### MLOps Tools and Implementation
- With MLOps we use *contianers*, that's where **Docker** comes in. *Containers* as a sealed package will contain the model and everything it needs to run. For our Fraud Detecting system, the Data Scientist would package their python model in a docker container. This container can run anywhere, either laptop or on the bank servers, it will work exactly the same way.
- For Production we need to make sure our models run reliably at scale, for that we deploy our model containers on **Kubernetes** *cluster*, such as *EKS* we solve environment consistency problem. Kubernetes then handles all the heavy lifting of scaling our fraud detection service up during, scaling up and down as per the requirement of the situation. 
- The best part is we can define this entire setup as *"Infrastructure as Code"*, *IaC* using tools like **Terraform** or **AWS CloudFormation**. This means that entire MLOps pipeline from development environment to production is documented, version control and reproducible. No more suprises when moving develpment to production.
- Fixing Performance Issues: MLOps includes automated testing. This is part of CI/CD or continous improvement/ continous deployment pipelines. Before the model goes live, the system or pipeline runs different types of tests(Unit/Performance Tests) to check if the model is fast enough.
- So, they will do speed testing of the model. The CI/CD pipelines runs a test that shows that the model is too slow. The team will fix it before it gets released and they could now optimize the model or use techniques likes batch processing, etc, to speed it up. 
- MLOps also addresses the data drift in an efficient way. MLOps let's you see new upcoming data. It uses data validation tools like **TenserFlow**, **Data Validation** or **Great Expecations** to compare what's coming in now with what the model was trained on. If the new data looks different, with above data monitoring tools(**TenserFlow**, etc.), it will be noticed quickly and they could alert the team or even automatically retrain the model with examples of the new fraud types.
- With MLOps everything is recorded automatically using experiment tracking tools like **MLFlow** or **DVC**, data version control, so, you have the exact data for training, all the steps used to clean and prepare the data, all the models settings and parameters results of each training attempt, and so on. These tracking tool means that the team could easily see exactly how the original model was built and recreate that perfectly with additional new fraud examples.
- MLOps involes setting up dashboard as well using tools like **Prometheus** & **Grafana** that shows how accurate the model is right now, how many transaction it's processing and how fast, if it's flagging too many legitimate transactions, whether it's missing actual frauds and so on. 

*Our banking team will see this metrics in real time or get instant alerts, let's say the fraud detection rate drops below an acceptable level, so that they can take proactive action to fix it. So that will be our continous proper monitoring of the running model in production*.

## MLOps Workflow
- Data Engineers set up data pipelines using tools like **Airflow** or **Kubeflow** that collect and clean transaction data on a daily basis.
- Data Scientist build models in standardized environments using **Docker** *contianers* and not in some custom environment in their laptops but in an environment which is very much similar to production environment.
- When th new model is ready for deployment. CI/CD tools like **Jenkins**, **GitLab CI**, **GitHub Actions** run automated tests to check the accuracy and speed for models with Unit Tests, Integration Tests, Performance Tests before deployment is allowed, if all tests pass the model is deployed automatically to a stagging environment, maybe a **Kubernetes** *cluster* where additional tests run with real transaction data before it can be deployed to final production environment.
- If the new version of fraud detection Model demonstrates better performance such as catching more fraud or flagging less false alarms, the deployment tool like **Argo CD** or **GitLab CI** or **GitHub Actions** can automatically replace the older production model running in the cluster with the new one without a downtime. 
- After it gets deployed to production the monitoring tools like **Prometheus** or **Grafana** will track how well the model is working 24/7. It is important especially within first hours or days of deploying a new version to immediately catch any issues that might slip through all these automated tests into the production. 
- We also have ongoing monitoring for performance degradation of the model because of changing data or changing customer behavior which again is a phenomenon known as model drift. So, tools like **Prometheus** automate alerting when metrics deviate from expected levels, which enables fast response. 
- So, when new fraud patterns emerge, the monitoring system alerts the team that the model needs to be updated.

*Some advanced systems can even trigger automated steps. So, if the fraud detection model starts missing new types of fraud or starts making more mistakes, the system monitoring notices this drop in accuray and it can actually automatically kick off a task to retrain the model with fresh examples. So the model learns to catch new fraud quickly and this way the bank stays protected without waiting for manual fixes*.

MLOps bridges the gap Experimental Machine Learning and prodcution-ready AI systems. MLOps makes it easier for teams to build, test and deploy machine learning systems that work in the real world. Whether you are creating models, deploying them or managing  the teams and infrastructure to support them. MLOps skills are becoming pretty demanded, important and valuable in this AI landscape.
